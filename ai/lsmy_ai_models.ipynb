## 1. Mục tiêu
Phát triển các mô hình học sâu - Deep Learning kiến trúc nhẹ, tối ưu cho xử lý tại biên (Edge Computing) trên thiết bị nhúng chạy Yocto Linux:
* **Human Counter:** Đếm số lượng người trong khu vực camera thời gian thực.
* **Behavior Detection:** Nhận diện tình trạng mệt mỏi (ngáp, gục đầu) hoặc mất tập trung.
* **Safety Compliance:** Mở rộng phát hiện đeo khẩu trang và tụ tập đông người.

## 2. Quy trình phát triển

### Chặng 1: Dữ liệu
* **Nguồn:** Dữ liệu từ Google Drive (được Mount vào `/content/drive`).
* **Xử lý:** Augmentation để thích ứng với ánh sáng phòng Lab.

### Chặng 2: Huấn luyện
* **Kiến trúc:** Sử dụng **MobileNetV2** hoặc **YOLOv8-tiny** để tối ưu cho phần cứng yếu.
* **Kỹ thuật:** Transfer Learning từ bộ trọng số ImageNet/COCO.

### Chặng 3: Tối ưu & Chuyển đổi
* **Quantization:** Chuyển đổi sang **INT8/Float16** để giảm 4 lần dung lượng.
* **Export:** Xuất file `.tflite` để chạy trên phần cứng biên.
